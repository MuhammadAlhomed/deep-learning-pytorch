# Math numpy and Pytorch

```python
# import libraries
import numpy as np
import torch
```

## Using numpy

```python
# Create a vector
nv = np.array([[1,2,3,4]])
print(nv), print()

# transpose it
print(nv.T)
```

Notice how we used double `[]` like `[[]]`.

This is **important!**

```python
print(f'Single bracket shape: {np.array([1,2,3,4]).shape}')
print(f'Double bracket shape: {np.array([[1,2,3,4]]).shape}')
```

Double bracket produces far less issues generally.
- `[]` indicates this is "4 columns with no rows" 
- `[[]]` indicate this is a "single row with 4 columns".

```python
# repeat for matrix
nm = np.array([[1,2,3,4], [4,5,6,7]])
print(nm), print()

# transpose it
print(nm.T)
```

## Using pytorch

```python
# create a vector
tv = torch.tensor([ [1,2,3,4] ])
print(tv), print()

# transpose it
print(tv.T)
```

```python
# repeat for matrix
tm = torch.tensor([[1,2,3,4], [4,5,6,7]])
print(tm), print()

# transpose it
print(tm.T)
```

Examing the types

```python
# Examine types

print(type(nv))
print(type(nm))
print(type(tv))
print(type(tm))
```
