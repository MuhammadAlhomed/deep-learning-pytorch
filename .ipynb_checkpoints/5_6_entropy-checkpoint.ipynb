# Entropy

```python
# import libraries
import numpy as np
import matplotlib.pyplot as plt
```

## Reminder of entropy

$$H(p) = -\sum_x p(x)\log(p(x))$$

```python
# Probabilty of an event happening 
p = .25

# WRONG FORMULA!
H = - (p*np.log(p))
print('Wrong entropy: ' + str(H))
```

The formula above is wrong because we calculated **the chance of the event happeing** but NOT **the chance of the event not happening**

```python
# corrected formula
x = [.25, .75]

H = 0
for p in x:
    H -= p * np.log(p)

print('Correct entropy: ' + str(H))
```

```python
# also correct, written out for N=2 events
H = -( p*np.log(p) + (1-p)*np.log(1-p) )
print('Correct entropy: ' + str(H))
```

# Cross-Entropy

```python
# probabilities MUST sum to 1.0
p = [1, 0] 
q = [.25 , .75]

H = 0
for i in range(len(p)):
    H -= p[i] * np.log(q[i])

print('Cross entropy: ' + str(H))
```

```python
# loopless form
# also correct, written out for N=2 events
H = -( p[0]*np.log(q[0]) + p[1]*np.log(q[1]) )
print('Correct entropy: ' + str(H))
```

Notice how `p = [1, 0]`. the second term is `0`. which means the term will be cancelled. we can simplify it. also `1` can be removed because `n * 1 = n`.

```python
# simplification
H = -np.log(q[0])
print('Manually simplified: ' + str(H))
```

The formula we got now is considered **Binary Cross-Entropy**.

## Cross-entropy using pytorch

```python
import torch
import torch.nn.functional as F

# convert lists to tensors
p_tensor = torch.Tensor(p)
q_tensor = torch.Tensor(q)

# apply bce
F.binary_cross_entropy(p_tensor,q_tensor)
```

notice how the answer is different. its not `1.386...`. That's because pytorch made this function inverted.

It expects `(model_predicted_probability, label)`

since p is our labels and q is prediction it came out wrong. So let's correct it.

```python
F.binary_cross_entropy(q_tensor,p_tensor)
```
