# Mean & Variance

```python
# import libraries
import numpy as np
```

```python
# create a list of numbers to compute the mean and variance of
x = [1, 2, 4, 6, 5, 4, 0]
n = len(x)

# compute the mean
mean = np.mean(x) # or np.sum(x) / n
```

```python
# variance
var1 = np.var(x)
var2 = 1/(n-1) * np.sum( (x - mean)**2 )
print(var1)
print(var2)
```

**NOTE:** the numbers above are different because `np.var(x)` does not do `n-1` by default. This is called a **biased variance**. We can fix making it **unbiased** by adjusting the degrees of freedom.

```python
var1 = np.var(x, ddof=1)

print(var1)
print(var2)
```
