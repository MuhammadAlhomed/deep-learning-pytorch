# Min/Max, Argmin/Argmax

````python
# import libraries
import numpy as np
import torch
````

## Using numpy

````python
# create a vector

#   Indice    0   1  2   3
v = np.array([1, 40, 2, -3])

print(f'Min value: {np.min(v)}')
print(f'Max value: {np.max(v)}')
print(f'Argmin :   {np.argmin(v)}')
print(f'Argmax :   {np.argmax(v)}')
````

````python
# repeat with matrix

m = np.array([
    [0, 1, 10],
    [20, 8, 5]
    ])

print(f'Min of ENTIRE matrix:                {np.min(m)}')
print(f'Min of each column (across rows):    {np.min(m, axis=0)}')
print(f'Min of each row (across columns):    {np.min(m, axis=1)}')
print('---')
print(f'Argmin of ENTIRE matrix:             {np.argmin(m)}')
print(f'Argmin of each column (across rows): {np.argmin(m, axis=0)}')
print(f'Argmin of each row (across columns): {np.argmin(m, axis=1)}')
````

## Using Pytorch

````python
# create a vector
v = torch.tensor([1, 40, 2, -3])

print(f'Min value: {torch.min(v)}')
print(f'Max value: {torch.max(v)}')
print(f'Argmin :   {torch.argmin(v)}')
print(f'Argmax :   {torch.argmax(v)}')
````

````python
# repeat with matrix 

# ======================
# NOTE THIS IS DIFFERENT
# ======================

m = torch.tensor([
    [0, 1, 10],
    [20, 8, 5]
    ])

min0 = torch.min(m, axis=0)
min1 = torch.min(m, axis=1)

print(f'Min of ENTIRE matrix:                {torch.min(m)}')
print(f'Min of each column (across rows):    {min0}')
print(f'Min of each row (across columns):    {min1}')
````

Notice how we got 
```
torch.return_types.min(
    values=tensor([0, 5]),
    indices=tensor([0, 2]))
```
Pytorch returns a class with the values and indice. we need to access them differnetly.

````python
print(f'Min of ENTIRE matrix:                {torch.min(m)}')
print(f'Min of each column (across rows):    {min0.values}')
print(f'Min of each row (across columns):    {min1.values}')
print('---')
print(f'Argmin of ENTIRE matrix:             {torch.argmin(m)}')
print(f'Argmin of each column (across rows): {min0.indices}')
print(f'Argmin of each row (across columns): {min1.indices}')
````
