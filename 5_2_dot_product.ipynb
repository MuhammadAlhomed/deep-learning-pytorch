# Dot Product

```python
# import libraries
import numpy as np
import torch
```

## Using numpy

```python
# create a vector
nv1 = np.array([1,2,3,4])
nv2 = np.array([4,5,6,7])

# dot product via function
print(np.dot(nv1, nv2))

# dot product via computation
print(np.sum(nv1*nv2)) # * means element wise multiplication. Not a dot product. so we add np.sum() to the equation to make it a dot product.

# dot product via computation 2
print(nv1 @ nv2) # @ means matrix multiplication, both dot and cross product.
```

```python
## Using pytorch
tv1 = torch.tensor([1,2,3,4])
tv2 = torch.tensor([4,5,6,7])

# dot product via function
print(torch.dot(tv1, tv2))

# dot product via computation
print(torch.sum(tv1*tv2)) 

# dot product via computation 2
print(tv1 @ tv2) # @ means matrix multiplication, both dot and cross product.
```
