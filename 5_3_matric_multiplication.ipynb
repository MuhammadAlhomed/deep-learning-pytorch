# Matrix Multiplication (Dot Product)

```python
# import libraries
import numpy as np
import torch
```

## Using numpy

```python
# create random matrix
A = np.random.randn(3,4)
B = np.random.randn(4,5)

print(A @ B) # result shape (3,5)
```

## Using pytorch

```python
# create random matrix
A = torch.randn(3,4)
B = torch.randn(4,5)

# Alternative way to make random matrix. np.ndarray -> tensor
C = np.random.randn(4,7)
C = torch.tensor(C, dtype=torch.float)

print(A@B) # result shape (3,5)
print(A@C) # result shape (3,7)
```
