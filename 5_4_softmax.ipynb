# Softmax

```python
# import libraries
import numpy as np
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
```

## Using numpy (manually)

```python
# manually implementing softmax

z = [1,2,3]

# compute softmax result
num = np.exp(z)
denominator = np.sum(np.exp(z))
sigma = num / denominator

print(sigma) # list of "probabilities"
print(np.sum(sigma)) # sum of "probabilities", equals 1.0
```

```python
# repeat with some random integers
z = np.random.randint(low=-5, high=15, size=25)
print(z)

# compute softmax result
num = np.exp(z)
denominator = np.sum(np.exp(z))
sigma = num / denominator

# plot the result
plt.plot(z, sigma, 'ko')
plt.xlabel('Original number (z)')
plt.ylabel('Softmaxed $\sigma$')
plt.yscale('linear') # linear or log
# plt.title('$\sum\sigma$ = %g' %np.sum(sigma)) # oldstyle formatting
plt.title(f'$\sum\sigma$ = {np.sum(sigma)}:g') # new formatting
plt.show()
```

## Using Pytorch

```python
# we will use pytorch.nn module

softmaxer = nn.Softmax(dim=0)

# then apply the data to that function
sigmaTensor = softmaxer(torch.Tensor(z))

print(sigmaTensor)
```

```python
# show that they are the same
plt.plot(sigma, sigmaTensor, 'ko')
plt.xlabel('"Manual" softmax')
plt.ylabel('Pytorch nn.Softmax')
plt.title(f'The two methods corrolate at r={np.corrcoef(sigma, sigmaTensor)[0,1]}')
plt.show()
```
